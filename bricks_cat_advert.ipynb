# Bricks Pet Food Advert — Image Combination

Uses the **Edit Images** functionality from `image_gen.py` (the *Combining several images* pattern) to merge the Bricks pet food bag and the grey cat into a single, natural-looking advertisement.

```python
from dotenv import load_dotenv
from openai import AsyncOpenAI
from image_gen import ImageGen
import os

load_dotenv(override=True)

client = AsyncOpenAI(
    api_key=os.getenv("DATABRICKS_TOKEN"),
    base_url=f"{os.getenv('DATABRICKS_HOST')}/serving-endpoints",
)
gen = ImageGen(client)
```

## Preview source images

```python
source_images = ["input/bricks.png", "input/grey_cat.jpg"]
gen.show(source_images)
```

## Generate the advert

Combines the Bricks pet food bag (branding, product) with the grey British Shorthair cat into a polished, studio-quality advertisement.

```python
prompt = """
Create a high-quality, professional pet food advertisement that naturally combines both images.

The scene: the grey British Shorthair cat sits beside the Bricks Premium Blend Pet Food bag on a
clean, warmly lit surface. The cat looks content and healthy, glancing toward the bag with
relaxed, curious eyes. The Bricks bag is fully visible and prominently featured, with all its
original branding intact (logo, 'HIGH QUALITY', 'PET FOOD', 'PREMIUM BLEND', net weight).

Lighting: warm, natural studio light, soft shadows. Background: a clean, blurred warm-neutral
backdrop that complements the bag's earthy orange-and-cream palette.

The overall look is a polished commercial photograph you would see in a pet store or magazine —
photorealistic, well-composed, inviting. No text overlays, no watermarks.
"""

edits = await gen.edit(
    prompt=prompt,
    images=source_images,
    file_prefix="bricks_cat_advert",
    quality="high",
    size="1024x1536",
    input_fidelity="high",
    n=2,
)
```
